# Smart Building Pack - Weather Gateway (MQTT)

This notebook shows how to:
1. Connect to SPX with `spx-python`
2. Check the instance state
3. Update `k__outdoor_temperature_c` and `k__brightness_lux` on the weather gateway instance

Prerequisites:
- SPX server running at `http://localhost:8000`
- Smart Building Pack installed (profile `bms_quickstart`)
- `SPX_PRODUCT_KEY` set in your environment
- Instance key: `spx_weather_gateway_wago_pfc200_vaisala_wxt530_mqtt`


```python
import os
import spx_python

SPX_BASE_URL = os.environ.get("SPX_BASE_URL", "http://localhost:8000")
PRODUCT_KEY = os.environ.get("SPX_PRODUCT_KEY")
INSTANCE_KEY = "spx_weather_gateway_wago_pfc200_vaisala_wxt530_mqtt"

if not PRODUCT_KEY:
    raise ValueError("SPX_PRODUCT_KEY is required.")

client = spx_python.init(address=SPX_BASE_URL, product_key=PRODUCT_KEY)
print(f"Connected to {SPX_BASE_URL}")
```

```python
try:
    instance = client["instances"][INSTANCE_KEY]
except Exception as exc:
    raise RuntimeError(
        f"Instance {INSTANCE_KEY!r} not found. Start the Smart Building Pack first."
    ) from exc

state = instance.state
print(f"Instance state: {state}")
if (state or "").lower() != "running":
    instance.start()
    print("Instance started.")
```

```python
attrs = instance["attributes"]
outdoor_temp = attrs["k__outdoor_temperature_c"]
brightness = attrs["k__brightness_lux"]

print("Before update:")
print("  k__outdoor_temperature_c =", outdoor_temp.external_value)
print("  k__brightness_lux =", brightness.external_value)

outdoor_temp.internal_value = 12.5
brightness.internal_value = 15000.0

print("After update:")
print("  k__outdoor_temperature_c =", outdoor_temp.external_value)
print("  k__brightness_lux =", brightness.external_value)
```
